In Python you have to follow a few rules when you give names to your variables:

1. Variable names have to start with a letter:
    * `my_variable`: Valid
    * `2no`: Not valid!

2. The rest of the name can contain letters or numbers. You can use an underscore if you want a name made of several words:
    * `variable2`: Valid
    * `hey@no`: Not valid!
    * `hey_no`: Valid

3. Keep in mind that variable names are case-sensitive
    * `Yes` and `yes` are two different variables.

4. Important: each variable name should be unique, otherwise, if you use it again, the value will be overwritten

````python
a = 5
b = 1

a = 6
````

## Accessing values in a list


The elements of a list are accessed by their index, that is, their position in the list. Keep in mind that in Python the first position is ``0``!

````python
students = ["Julian", "Marcela", "Sebastian"]
````

````python
students[0]
````

````python
students[1]
````

````python
students[2]
````

You can also go through the list backwards, from last to first. To do this you use negative numbers, where ``-1`` is the last position, ``-2`` the second to last, and so on.

````python
students[-1]
````

To check whether an element is in a list we can use ``in``.

``` python

```

````python
print("Juan" in students)
````

````python
print("Marcela" in students)
````

````python
print("sebastian" in students)
````

**Adding elements to a list:**
* ``append``: add a single new element at the end of the list
* ``extend``: add the elements of another list at the end of the list
* ``insert``: add an element at the position you specify

````python
# append
dishes = ['arepa', 'sancocho', 'arroz', 'sopa']
dishes.append('lentejas')
print(dishes)
````

````python
dishes = ['arepa', 'sancocho', 'arroz', 'sopa']
dishes.append(['lentejas', 'frijoles'])
print(dishes)
````

````python
# extend
dishes = ['arepa', 'sancocho', 'arroz', 'sopa']
dishes.extend(['lentejas', 'frijoles'])
print(dishes)
````

````python
dishes = ['arepa', 'sancocho', 'arroz', 'sopa', 'lentejas', 'frijoles']
dishes.insert(2, "flan")
print(dishes)
````

**Removing elements from a list:**
* ``clear``: removes every element from the list
* ``pop``
    - Removes the element at the position you specify and returns it.
    - If no position is given, it removes the last element and returns it
* ``del``: removes the element at the position you specify

````python
#clear
dishes = ['arepa', 'sancocho', 'arroz', 'sopa', 'lentejas', 'frijoles']
dishes.clear()
print(dishes)
````

````python
# pop
dishes = ['arepa', 'sancocho', 'arroz', 'sopa', 'lentejas', 'frijoles']
dishes.pop()
````

````python
print(dishes)
````

````python
# pop
dishes = ['arepa', 'sancocho', 'arroz', 'sopa', 'lentejas', 'frijoles']
dishes.pop(2)
````

````python
print(dishes)
````

````python
# del
dishes = ['arepa', 'sancocho', 'arroz', 'sopa', 'lentejas', 'frijoles']
del dishes[3]
print(dishes)
del dishes[1]
print(dishes)
````

````python
#del
dishes = ['arepa', 'sancocho', 'arroz', 'sopa', 'lentejas', 'frijoles']
del dishes[6]
print(dishes)
````

**Other useful methods:**
* `count`: returns the number of times an element is found in the list.

# Dictionaries

> A dictionary lets you store key, value pairs. The key is used to look up the value. This is very useful when you want to find elements by their key (like a traditional dictionary).

- Dictionaries are created using curly brackets ``{key: value}``
- Values are accessed by their key, written in a way similar to lists: dictio[key] --> value
- The keys in a dictionary are unique!

````python
# define dictionary
cities = {
    'Medellin': 13350000, # key value pair
    'Cartagena': 18550000,
    'Ibague': 8400000,
    'Bogota': 1837442,
}

# display dictionary
print(cities)
````

To access one specific value:

````python
print(cities['Cartagena'])
````

To change or assign a value:

````python
cities["Copenhagen"] = 2389212
````

````python
print(cities)
````

Dictionaries let you build **more complex structures**, for example, structures where the value is **a list**:

````python
microbes = {"bacteria": ["Streptococcus pneumoniae", "Staphylococcus aureus"], 
            "fungi": ["Aspergillus versicolor", "Candida albicans"]}
print(microbes)
````
