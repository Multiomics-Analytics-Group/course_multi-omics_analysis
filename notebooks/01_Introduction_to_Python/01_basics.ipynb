## 1. Variables

A variable is a name for a value. You write `name = value`, and from then on Python
substitutes the value wherever it sees the name. That is the whole mechanism, and it is
the reason a script can be read: `min_completeness = 0.7` tells the next person what the
number 0.7 *means*, which `0.7` sitting on its own never does.

The `=` symbol is doing assignment, not stating a fact. It is an instruction — "put this
value in that name" — and it always runs right to left. We will meet the *other* equals,
`==`, which asks a question instead, in the next section.

Run the five cells below one at a time. The first three produce no output at all, because
storing something is silent; only the last two, which end in an expression rather than an
assignment, print anything.

````python
my_first_variable = 5
````

````python
my_second_variable = 12
````

````python
c = "Where, "
d = "when, "
e = "why?"
````

````python
my_first_variable + my_second_variable
````

````python
c + d + e
````

Notice that `+` did two different jobs there: it added two numbers, and it glued three
pieces of text together end to end. Python decides what an operator means from the kind of
thing it is applied to — which is the subject of section 2, and the source of one of the
error messages in section 4.

### Naming rules

Python is strict about names, and the rules are worth learning now rather than by
accident:

1. A name has to start with a letter (or an underscore):
    * `my_variable`: valid
    * `2no`: not valid!

2. The rest of the name can contain letters, numbers and underscores. Use underscores to
   join several words — that is the convention in Python, and it is what makes
   `protein_groups_matrix` readable:
    * `variable2`: valid
    * `hey@no`: not valid!
    * `hey_no`: valid

3. Names are **case-sensitive**. `Yes` and `yes` are two different variables, and mixing
   them up is a classic way to spend twenty minutes hunting a bug that is not there.

4. A name holds one value at a time. Assign to it again and the old value is simply gone —
   Python will not warn you.

That last rule is the one that bites in a notebook, where you can run cells in any order.
Read the cell below and decide what you expect before you run it.

````python
a = 5
b = 1

a = 6
````

What value does `a` have now? There is no history and no undo: the second assignment
overwrote the first, and `a` is whatever the most recently executed assignment put there.

> ⚙️ This is why a notebook that runs correctly top to bottom can still be wrong when you
> have been jumping around in it. Before you trust a result, restart the kernel and run
> every cell in order. It takes thirty seconds and it catches an entire class of mistake.

## 2. Types

Every value has a **type**, and the type decides what you are allowed to do with it. This
is not bureaucracy — it is the difference between `2 + 3` giving 5 and `"2" + "3"` giving
`"23"`. Python will happily do both, and it will not ask which one you meant.

Four types cover almost everything in this course:

| Type | Python name | Example | What it is for |
|---|---|---|---|
| Integer | `int` | `45` | counts — patients, proteins, positions in a list |
| Float | `float` | `1.23` | measurements — intensities, p-values, fold changes |
| String | `str` | `"CRKP"` | text — sample identifiers, gene names, group labels |
| Boolean | `bool` | `True` | answers to yes/no questions — the result of a comparison |

In the protein matrix, the intensities are floats, the protein accessions are strings, and
the "is this protein missing?" mask that we build on Day 2 is an entire table of booleans.

### Integer — a whole number

````python
a = 123
````

### Float — a decimal number

Floats are stored approximately, which is why `0.1 + 0.2` is famously not exactly `0.3`.
It never matters for intensities; it matters a great deal if you ever test two measured
numbers for exact equality.

````python
b = 1.23
````

### String — text

Quotes are what make text text. Single and double quotes behave identically, so pick one

````python
c = "Hello world!"
````

### Boolean — true or false

`True` and `False` are values in their own right, spelt with a capital letter and written
without quotes. You rarely type them; you get them back from questions you ask the data —
"is this value missing?", "is this p-value below 0.05?" — and then use them to select rows.

````python
d = True
````

````python
e = False
````

a float at all. It refuses far more readily across types that have nothing in common,
as we are about to see.

````python
a == b
````

## 3. Functions and methods

A function is a named piece of work that somebody has already done. You hand it some
values — its **arguments**, in the brackets — and it hands one back. `sum` adds up a
collection of numbers; `print` displays something; `len`, later on, counts elements. You
do not need to know how any of them are implemented, which is exactly the point: functions
let you use an idea without re-deriving it.

The brackets are not decoration. `sum` on its own is the function itself; `sum([a, b])`
is the function *being called*. Forgetting the brackets is one of the most common
beginners' mistakes, and it fails silently rather than loudly.

A **method** is a function that belongs to a particular value, and it is written after a
dot: `my_list.append(...)`, `"text".upper()`. Same idea, different punctuation. Most of
what you write in pandas is methods on a table.

Below, `sum([a, b])` adds our integer and our float, and `print(c)` shows the string. Look
closely at what `c` contains: we set it to `"Where, "` back in section 1 and then to
`"Hello world!"` a moment ago. The second assignment won.

````python
sum([a,b])
````

````python
print(c)
````

### Writing your own

underneath. `return` says what comes back out; a function without a `return` still does its
work — `shout_hello` prints — but hands back nothing you can store.

The indentation is not style. In Python, indentation *is* the syntax that says "this line
belongs inside that block", and getting it wrong is an error rather than an untidiness.

````python
def add_two_numbers(a, b):
    return a + b
````

````python
def shout_hello():
    print("Hello!!!!!!")
````

## 4. Understanding error messages

Errors are not a sign that you are bad at this. They are the normal medium of programming:
experienced analysts produce them constantly and simply read them faster. Python is being
helpful when it stops — the alternative, quietly carrying on with something you did not
mean, is far worse and is how wrong results get published.

Almost every error falls into one of three families. Either we referred to something that
does not exist, or we asked for something that is not allowed for that type, or we wrote
something Python cannot parse at all.

***

The message itself has a shape worth learning. The block of text is a **traceback**: it
shows where Python was when things went wrong, reading from the outside in. The line that
matters is **the last one** — that is the actual error and its type, and it is where you
should look first, every time.

```python
---------------------------------------------------------------------------
NameError                                 Traceback (most recent call last)
Cell In[15], line 1
----> 1 my_variable   # The arrow points to the line where the error happens

NameError: name 'my_variable' is not defined    # the last line always shows the error and its type
```

> ⚙️ Read the last line, then the arrow, then stop. Beginners scroll up in alarm and read
> the middle, which is the least informative part. If the last line still means nothing,
> paste it into a search engine — you will not be the first person to have seen it.

The next four cells are **supposed to fail**. Run them one at a time and read what comes
back; each one demonstrates a different family of mistake.

**`ZeroDivisionError`** — the arithmetic is well formed but has no answer. Python raises
rather than returning infinity or `NaN`, which is the right call: in a real analysis, a
division by zero usually means a denominator you assumed was non-zero is empty, and you
want to know that.

````python
1 / 0
````

**`NameError`** — we used a name that has never been assigned. In a notebook this most
often means the cell that defines it has not been run yet, or was run and then edited.
Nine times out of ten the fix is to re-run the earlier cell, not to change this one; the
tenth time it is a typo in the name.

````python
my_variable
````

**`TypeError`** — the operation is fine, but not for these types. `+` concatenates two
strings and adds two numbers; it has no meaning for one of each, and Python refuses to
guess whether you wanted `"balance12"` or an arithmetic error. This is the type system
from section 2 doing its job. The fix is to convert deliberately: `"balance" + str(12)`.

````python
"balance" + 12
````

**`SyntaxError`** — this one is different in kind. The three above happened *while* the
code ran; this one happens before anything runs at all, because Python cannot even work
out what the line says. `23andme` breaks naming rule 1 from section 1 — a name may not
start with a digit. Whenever a cell fails without producing any of its normal output,
suspect a syntax error, and look at the line *above* the one Python points at, since an
unclosed bracket is only noticed on the following line.

````python
23andme = 12
````

To summarise the four, and two more you will certainly meet later today:

| Error | What it usually means | First thing to check |
|---|---|---|
| `NameError` | the name does not exist | did you run the cell that defines it? |
| `TypeError` | wrong type for this operation | what does `type()` say it is? |
| `SyntaxError` | Python cannot read the line | brackets, quotes, colons, the line above |
| `ZeroDivisionError` | a denominator was zero | which value did you assume was non-zero? |
| `IndexError` | asked for a position that is not there | how long is the list *really*? |
| `KeyError` | asked for a key that is not there | spelling and case of the key |

The last two belong to lists and dictionaries, and we will trigger an `IndexError` on
purpose in section 8.

## 8. Lists

A list holds several values in a definite order. That order is the point: a list of sample
identifiers and a list of their group labels line up position by position, and every
matrix we touch is ultimately a list of rows, each row a list of numbers. When we transpose
the protein matrix on Day 2 so that samples are in rows, we are rearranging exactly this.

Lists are built with square brackets `[]` or with `list()`, they can be joined with `+`
like strings, and they can hold anything — numbers, text, or other lists. The methods you
will use most:

| Method   |  Effect                                  |
| :-----:  |  :----------:                                  |
| append() | Adds a single element to the end of the list        |
| extend() | Adds the elements of another list to the end of the list, one after the other  |
| remove() | Removes an element from the list                 |
| clear()  | Empties the list              |
| copy()   | Returns a copy of the list                    |
| sort()   | Sorts the list                               |
| reverse()| Reverses the order of the list       |

One thing to notice as you go: most of these change the list **in place** and return
nothing. `my_list.sort()` sorts the list and hands back `None` — so `x = my_list.sort()`

in the language, and knowing it now saves an afternoon later.

**Example.** A list of five elements:

### Adding elements

Three methods, and the difference between the first two matters:

* ``append``: adds **one** element at the end — even if that element is itself a list, in
  which case you get a list nested inside your list
* ``extend``: adds the elements **of** another list, one after the other, keeping the list flat
* ``insert``: adds an element at the position you specify, pushing the rest along

Compare the second and third cells below. `append(['lentejas', 'frijoles'])` gives a list
of five things, the last of which is a list; `extend(['lentejas', 'frijoles'])` gives a
flat list of six. Accidentally nesting a list is a common cause of a `TypeError` several
steps later, far from where the mistake was made.

````python
# append
dishes = ['arepa', 'sancocho', 'arroz', 'sopa']
dishes.append('lentejas')
print(dishes)
````

````python
dishes = ['arepa', 'sancocho', 'arroz', 'sopa']
dishes.append(['lentejas', 'frijoles'])
print(dishes)
````

````python
# extend
dishes = ['arepa', 'sancocho', 'arroz', 'sopa']
dishes.extend(['lentejas', 'frijoles'])
print(dishes)
````

````python
dishes = ['arepa', 'sancocho', 'arroz', 'sopa', 'lentejas', 'frijoles']
dishes.insert(2, "flan")
print(dishes)
````

## 10. Dictionaries

A list finds things by *position*. A dictionary finds them by *name*.

> A dictionary lets you store key–value pairs. The key is used to look up the value, which
> is very useful when you want to find elements by their name rather than by counting to
> them (like a traditional dictionary).

- Dictionaries are created with curly brackets, ``{key: value}``
- Values are read with square brackets, like lists, but using the key: `dictio[key]`
- Keys are **unique** — assigning to an existing key replaces its value

This is the right structure whenever position is meaningless and identity is everything,
which describes most metadata. Asking for `metadata["KP07"]` is both clearer and safer than
remembering that patient KP07 happens to be row 23; rows move, names do not.

````python
# define dictionary
cities = {
    'Medellin': 13350000, # key value pair
    'Cartagena': 18550000,
    'Ibague': 8400000,
    'Bogota': 1837442,
}

# display dictionary
print(cities)
````

To read one specific value, give its key. Ask for a key that is not there and you get a
`KeyError` — the dictionary's version of the `IndexError` we met above.

````python
print(cities['Cartagena'])
````

To change a value, or add a new pair, assign to the key. There is no separate "add" method:
the same line does both, depending on whether the key already exists. That convenience is
also a trap — a typo in a key silently creates a new entry rather than raising an error.

````python
cities["Copenhagen"] = 2389212
````

````python
print(cities)
````


A **dictionary of dictionaries** — one row per patient, one named column per fact. Read
`metadata["KP7"]["group"]` aloud and it says exactly what it does. (The identifiers really
do look like this: `Con1`–`Con15`, `KP1`–`KP15` for the *susceptible* infections, and
`CRKP1`–`CRKP15` for the resistant ones. The `KP` prefix meaning CSKP is a trap worth
remembering.) This is precisely what
`sample_metadata.tsv` contains for our 45 patients, and when pandas loads that file after
the break it is building this structure and giving it a nicer interface.

> 🧬 Keeping the measurements and the metadata in **separate** objects, joined on an
> explicit sample identifier, is the habit worth taking from this hour. Encoding a
> patient's group in the column name of a matrix — or in a file name — is how cohorts get
> scrambled, and it is a mistake that is almost impossible to detect afterwards.
